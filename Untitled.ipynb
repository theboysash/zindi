```python
from pathlib import Path

BASE = Path('.')

TRAIN_FILE      = str(BASE / 'train.csv')
VAL_FILE        = str(BASE / 'val.csv')
TEST_FILE       = str(BASE / 'test.csv')
SAMPLE_SUB_FILE = str(BASE / 'SampleSubmission.csv')
SUBMISSION_FILE = str(BASE / 'submission.csv')

ID_COL      = 'ID'
TARGET_COLS = [
    'no_water_access', 'no_sanitation_access', 'no_refuse_access',
    'no_energy_access', 'no_education_access',
]

RANDOM_STATE = 42
print('Config loaded')
```

```python
train      = pd.read_csv(TRAIN_FILE)
val        = pd.read_csv(VAL_FILE)
test       = pd.read_csv(TEST_FILE)
sample_sub = pd.read_csv(SAMPLE_SUB_FILE)

print(f'Train : {train.shape}')
print(f'Val   : {val.shape}')
print(f'Test  : {test.shape}')
```

```python
def engineer_features(df):
    df = df.copy()
    
    # ── Ordinal bands → integers ──────────────────────────────────────────────
    df['asset_index']      = df['asset_index'].map(
        {'0-2': 0, '3-4': 1, '5-6': 2, '7-8': 3, '9+': 4})
    df['debt_count']       = df['debt_count'].map(
        {'0': 0, '1': 1, '2': 2, '3+': 3})
    df['income_src_count'] = df['income_src_count'].map(
        {'0': 0, '1': 1, '2+': 2})
    df['hazard_count']     = df['hazard_count'].map(
        {'0': 0, '1': 1, '2': 2, '3+': 3})
    df['amenity_access']   = df['amenity_access'].map(
        {'0-2': 0, '3-4': 1, '5-6': 2, '7+': 3})
    df['chronic_count']    = df['chronic_count'].map(
        {'0': 0, '1': 1, '2+': 2})
    df['hh_size_band']     = df['hh_size_band'].map(
        {'1': 1, '2': 2, '3-4': 3, '5-6': 5, '7+': 7})
    df['rooms_band']       = df['rooms_band'].map(
        {'1-2': 1, '3-4': 3, '5+': 5})
    df['children_band']    = df['children_band'].map(
        {'0': 0, '1': 1, '2': 2, '3+': 3})
    df['age_band']         = df['age_band'].map(
        {'<25': 0, '25-34': 1, '35-44': 2, '45-54': 3,
         '55-64': 4, '65+': 5, '__SUPPRESSED__': np.nan})
    df['tenure']           = df['tenure'].map(
        {'Owned': 0, 'Renting': 1,
         'Allowed to stay rent free by owner': 2,
         'Squatting or living rent-free in an informal dwelling youve built, or in a vacant building or on vacant land': 3})
    df['transport_mode']   = df['transport_mode'].map(
        {'Car as driver': 0, 'Car as passenger': 1,
         'Minibus Taxi': 2, 'Walk': 3})
    df['health_status']    = df['health_status'].map(
        {'Excellent': 0, 'Good': 1, 'Poor': 2})
    df['marital']          = df['marital'].map(
        {'Married/common law marriage': 0,
         'In a relationship but not married': 1,
         'Single': 2, 'Divorced': 3, 'Widowed': 4})

    # ── Binary columns → 0/1 ─────────────────────────────────────────────────
    for col in ['internet', 'employed', 'social_grant', 'medical_aid', 'crime_victim']:
        df[col] = (df[col] == 'Yes').astype(int)

    df['food_insecure']    = (df['food_insecurity'] == 'insecure').astype(int)
    df['is_informal']      = (df['dwelling_type'] == 'Informal').astype(int)
    df['is_female']        = (df['sex'] == 'Female').astype(int)
    df['is_head']          = (df['head_household'] == 'Respondent').astype(int)
    df['is_black_african'] = (df['pop_group'] == 'Black African').astype(int)
    df['is_white']         = (df['pop_group'] == 'White').astype(int)

    # Drop original string columns now replaced
    df = df.drop(columns=['dwelling_type', 'food_insecurity', 'sex',
                           'head_household', 'pop_group'])

    # ── Composite scores ──────────────────────────────────────────────────────
    hh = df['hh_size_band'] + 1

    df['informal_score'] = (
        df['is_informal'] +
        (df['tenure'] >= 2).astype(int)
    )
    df['ses_vulnerability'] = (
        (df['employed'] == 0).astype(int) +
        (df['medical_aid'] == 0).astype(int) +
        (df['internet'] == 0).astype(int) +
        df['social_grant'] +
        (df['asset_index'] <= 1).astype(int) +
        df['food_insecure'] +
        (df['income_src_count'] == 0).astype(int)
    )
    df['crowding_proxy']    = df['hh_size_band'] / (df['rooms_band'] + 1)
    df['dependents_total']  = df['children_band'] + df['elderly_band']
    df['dependency_burden'] = df['dependents_total'] / hh
    df['isolation_score']   = (
        (df['internet'] == 0).astype(int) +
        (df['transport_mode'] >= 2).astype(int) +
        (df['amenity_access'] <= 1).astype(int) +
        (df['medical_aid'] == 0).astype(int)
    )
    df['health_burden'] = (
        df['chronic_count'] +
        df['hazard_count'] +
        df['health_status']
    )

    # ── Per capita features ───────────────────────────────────────────────────
    df['assets_per_person']         = df['asset_index'] / hh
    df['debt_per_person']           = df['debt_count'] / hh
    df['income_sources_per_person'] = df['income_src_count'] / hh
    df['rooms_per_person']          = df['rooms_band'] / hh

    # ── Interaction terms ─────────────────────────────────────────────────────
    df['informal_x_ses']   = df['informal_score'] * df['ses_vulnerability']
    df['asset_x_formal']   = df['asset_index'] * (1 - df['is_informal'])
    df['asset_x_informal'] = df['asset_index'] * df['is_informal']
    df['walk_x_informal']  = (df['transport_mode'] == 3).astype(int) * df['is_informal']
    df['crowded_informal'] = df['crowding_proxy'] * df['is_informal']
    df['ses_x_isolation']  = df['ses_vulnerability'] * df['isolation_score']
    df['female_informal']  = df['is_female'] * df['is_informal']

    return df

print('Feature engineering function defined')
```

```python
def add_extra_features(df):
    df = df.copy()

    # ── Team Feature 1: Triple Zero flag ─────────────────────────────────────
    # Worst case: no income + no assets + no amenities simultaneously
    df['triple_zero'] = (
        (df['income_src_count'] == 0).astype(int) &
        (df['asset_index'] == 0).astype(int) &
        (df['amenity_access'] == 0).astype(int)
    ).astype(int)

    # ── Team Feature 2: No Safety Net flag ───────────────────────────────────
    # Informal dwelling + not employed + no social grant
    df['no_safety_net'] = (
        (df['is_informal'] == 1).astype(int) &
        (df['employed'] == 0).astype(int) &
        (df['social_grant'] == 0).astype(int)
    ).astype(int)

    # ── Team Feature 3: Transport vulnerability ───────────────────────────────
    # Walk or Taxi = infrastructure vulnerable
    # transport_mode already encoded: Walk=3, Minibus Taxi=2
    df['is_transport_vulnerable'] = (df['transport_mode'] >= 2).astype(int)

    # ── Team Feature 4: Total vulnerable members ──────────────────────────────
    # Children + chronic conditions in one score
    df['total_vulnerable_members'] = (
        df['children_band'] + 
        df['chronic_count'] + 
        df['elderly_band']
    )

    # ── Midpoint refinement for hh_size and rooms ────────────────────────────
    # Use midpoints instead of lower bounds for better numeric representation
    df['hh_size_mid'] = df['hh_size_band'].map(
        {1: 1, 2: 2, 3: 3.5, 5: 5.5, 7: 9.0}
    )
    df['rooms_mid'] = df['rooms_band'].map(
        {1: 1.5, 3: 3.5, 5: 6.0}
    )

    # Recalculate crowding with midpoints — more accurate
    df['crowding_mid'] = df['hh_size_mid'] / (df['rooms_mid'] + 1)

    return df

print('Extra features function defined')
```

```python
# ── Feature Selection: Drop Redundant Features ────────────────────────────────
features_to_drop = [
    'crime_victim',              # near-zero variance
    'is_white',                  # exact mirror of is_black_african
    'debt_per_person',           # redundant with debt_count
    'dependency_burden',         # redundant with children_band
    'income_sources_per_person', # near-zero correlation + redundant
]

# Reload raw data to avoid any previous transformations
train = pd.read_csv(TRAIN_FILE)
val   = pd.read_csv(VAL_FILE)
test  = pd.read_csv(TEST_FILE)

X_train = add_extra_features(engineer_features(train.drop(columns=TARGET_COLS + [ID_COL])))
X_val   = add_extra_features(engineer_features(val.drop(columns=TARGET_COLS + [ID_COL])))
X_test  = add_extra_features(engineer_features(test.drop(columns=[ID_COL])))

y_train = train[TARGET_COLS].values.astype(int)
y_val   = val[TARGET_COLS].values.astype(int)

X_train = X_train.drop(columns=features_to_drop, errors='ignore')
X_val   = X_val.drop(columns=features_to_drop, errors='ignore')
X_test  = X_test.drop(columns=features_to_drop, errors='ignore')

print(f'X_train : {X_train.shape}')
print(f'X_val   : {X_val.shape}')
print(f'X_test  : {X_test.shape}')
print(f'\nNull counts in X_train:')
print(X_train.isnull().sum()[X_train.isnull().sum() > 0])
print(f'\nFinal features:')
print(list(X_train.columns))
```

```python
from lightgbm import LGBMClassifier, early_stopping, log_evaluation
from sklearn.model_selection import StratifiedKFold
import numpy as np

X_full = pd.concat([X_train, X_val], axis=0).reset_index(drop=True)
y_full = np.vstack([y_train, y_val])

strat_key = y_full.sum(axis=1)

N_FOLDS = 5
skf = StratifiedKFold(n_splits=N_FOLDS, shuffle=True, random_state=RANDOM_STATE)

oof_probs  = np.zeros_like(y_full, dtype=float)
test_probs = np.zeros((X_test.shape[0], len(TARGET_COLS)), dtype=float)
fold_scores = []

for fold, (train_idx, val_idx) in enumerate(skf.split(X_full, strat_key)):
    print(f'\n=== Fold {fold+1}/{N_FOLDS} ===')
    
    Xf_train, Xf_val = X_full.iloc[train_idx], X_full.iloc[val_idx]
    yf_train, yf_val = y_full[train_idx], y_full[val_idx]
    
    fold_test_probs = np.zeros((X_test.shape[0], len(TARGET_COLS)), dtype=float)
    
    for i, target in enumerate(TARGET_COLS):
        model = LGBMClassifier(
            n_estimators=1000,
            learning_rate=0.05,
            num_leaves=31,
            min_child_samples=20,
            class_weight='balanced',
            subsample=0.8,
            colsample_bytree=0.8,
            random_state=RANDOM_STATE,
            n_jobs=-1,
            verbose=-1
        )
        
        model.fit(
            Xf_train, yf_train[:, i],
            eval_set=[(Xf_val, yf_val[:, i])],
            callbacks=[
                early_stopping(100, verbose=False),
                log_evaluation(300)
            ]
        )
        
        oof_probs[val_idx, i] = model.predict_proba(Xf_val)[:, 1]
        fold_test_probs[:, i] = model.predict_proba(X_test)[:, 1]
        
    fold_preds = (oof_probs[val_idx] >= 0.5).astype(int)
    fold_f1    = f1_score(yf_val, fold_preds, average='macro', zero_division=0)
    fold_scores.append(fold_f1)
    print(f'  Fold {fold+1} macro-F1 @ 0.5: {fold_f1:.4f}')
    
    test_probs += fold_test_probs / N_FOLDS

print(f'\nMean CV macro-F1 @ 0.5: {np.mean(fold_scores):.4f} ± {np.std(fold_scores):.4f}')
```

```python
# Tune thresholds on out-of-fold predictions — unbiased estimate
best_thresholds = {}
tuned_f1s       = {}

print('Per-label threshold tuning on OOF predictions:\n')
print(f'  {"label":<25} {"thr":>5}  {"F1@0.5":>7}  {"F1@thr":>7}  {"delta":>7}')

for i, target in enumerate(TARGET_COLS):
    base_f1 = f1_score(y_full[:, i],
                       (oof_probs[:, i] >= 0.5).astype(int),
                       zero_division=0)
    best_f1, best_thr = 0, 0.5
    for thr in np.arange(0.05, 0.95, 0.01):
        preds = (oof_probs[:, i] >= thr).astype(int)
        f1    = f1_score(y_full[:, i], preds, zero_division=0)
        if f1 > best_f1:
            best_f1, best_thr = f1, thr
    
    best_thresholds[target] = round(best_thr, 2)
    tuned_f1s[target]       = best_f1
    print(f'  {target:<25} {best_thr:>5.2f}  {base_f1:>7.4f}  {best_f1:>7.4f}  {best_f1-base_f1:>+7.4f}')

macro_f1 = np.mean(list(tuned_f1s.values()))
print(f'\nOOF Tuned Macro-F1 : {macro_f1:.4f}')
print(f'Baseline was       : 0.4169')
print(f'Improvement        : +{macro_f1 - 0.4169:.4f}')
```

```python
submission = sample_sub[[ID_COL]].copy()

for target in TARGET_COLS:
    submission[target] = (
        test_probs[target] >= best_thresholds[target]
    ).astype(int)

# Sanity checks
assert list(submission.columns) == [ID_COL] + TARGET_COLS
assert len(submission) == len(sample_sub)
assert submission.isna().sum().sum() == 0
for col in TARGET_COLS:
    assert submission[col].isin([0, 1]).all()

print('All sanity checks passed.')
print(f'Submission shape: {submission.shape}')
print('\nPredicted positives per label:')
print((submission[TARGET_COLS].mean() * 100).apply(lambda x: f'{x:.1f}%'))

submission.to_csv(SUBMISSION_FILE, index=False)
print(f'\nSubmission saved to: {SUBMISSION_FILE}')
```

```python
from scipy.stats import spearmanr
import warnings
warnings.filterwarnings('ignore')

# ── Spearman Correlation: each feature vs each target ─────────────────────────
print("=== Spearman Correlation: Features vs Targets ===\n")

spearman_results = pd.DataFrame(index=X_train.columns, columns=TARGET_COLS, dtype=float)

for feat in X_train.columns:
    for j, target in enumerate(TARGET_COLS):
        corr, pval = spearmanr(X_train[feat], y_train[:, j], nan_policy='omit')
        spearman_results.loc[feat, target] = round(corr, 4)

print(spearman_results.to_string())

# Show top 10 most correlated features per target
print("\n=== Top 10 Features per Target (absolute Spearman) ===\n")
for target in TARGET_COLS:
    top = spearman_results[target].abs().sort_values(ascending=False).head(10)
    print(f"--- {target} ---")
    print(top.to_string())
    print()
```

```python
# ── Pearson Correlation: each feature vs each target ──────────────────────────
print("=== Pearson Correlation: Features vs Targets ===\n")

pearson_results = pd.DataFrame(index=X_train.columns, columns=TARGET_COLS, dtype=float)

for feat in X_train.columns:
    for j, target in enumerate(TARGET_COLS):
        corr = X_train[feat].corr(pd.Series(y_train[:, j]))
        pearson_results.loc[feat, target] = round(corr, 4)

print(pearson_results.to_string())

print("\n=== Top 10 Features per Target (absolute Pearson) ===\n")
for target in TARGET_COLS:
    top = pearson_results[target].abs().sort_values(ascending=False).head(10)
    print(f"--- {target} ---")
    print(top.to_string())
    print()
```

```python
# ── Inter-feature Correlation ─────────────────────────────────────────────────
# Find features that are highly correlated with each other
# These are redundant — keeping both adds noise without adding signal

corr_matrix = X_train.corr().abs()

# Get upper triangle only (avoid duplicates)
upper = corr_matrix.where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
)

# Find pairs above threshold
threshold = 0.85
high_corr_pairs = []
for col in upper.columns:
    for idx in upper.index:
        val = upper.loc[idx, col]
        if pd.notna(val) and val > threshold:
            high_corr_pairs.append((idx, col, round(val, 4)))

high_corr_pairs.sort(key=lambda x: x[2], reverse=True)

print(f"=== Highly Correlated Feature Pairs (r > {threshold}) ===\n")
if high_corr_pairs:
    print(f"{'Feature A':<30} {'Feature B':<30} {'Correlation':>12}")
    print("-" * 74)
    for a, b, r in high_corr_pairs:
        print(f"{a:<30} {b:<30} {r:>12}")
else:
    print("No pairs found above threshold.")

# Also show heatmap of inter-feature correlations
fig, ax = plt.subplots(figsize=(20, 16))
im = ax.imshow(corr_matrix.values, cmap='RdYlGn_r', vmin=0, vmax=1, aspect='auto')
ax.set_xticks(range(len(X_train.columns)))
ax.set_yticks(range(len(X_train.columns)))
ax.set_xticklabels(X_train.columns, rotation=90, fontsize=7)
ax.set_yticklabels(X_train.columns, fontsize=7)
ax.set_title('Inter-feature Correlation Matrix', fontsize=13, fontweight='bold')
plt.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
plt.tight_layout()
plt.savefig('inter_feature_correlation.png', dpi=120, bbox_inches='tight')
plt.show()
print('\nHeatmap saved.')
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.multioutput import MultiOutputClassifier

rf_model = MultiOutputClassifier(
    RandomForestClassifier(
        n_estimators=500,
        max_depth=None,
        min_samples_leaf=2,
        class_weight='balanced',
        random_state=RANDOM_STATE,
        n_jobs=-1
    )
)

rf_model.fit(X_train, y_train)

# Evaluate at 0.5
rf_preds = rf_model.predict(X_val)
rf_macro = f1_score(y_val, rf_preds, average='macro', zero_division=0)
rf_per_label = f1_score(y_val, rf_preds, average=None, zero_division=0)

print(f'RF Macro-F1 @ 0.5: {rf_macro:.4f}')
print(f'Baseline was      : 0.4169')
print()
print('Per-label F1:')
for col, score in zip(TARGET_COLS, rf_per_label):
    print(f'  {col:<25} {score:.4f}')

# Tune thresholds
rf_probs = np.column_stack([p[:, 1] for p in rf_model.predict_proba(X_val)])
best_thr_rf = {}
tuned_f1_rf = {}

print('\nThreshold tuning:')
for i, target in enumerate(TARGET_COLS):
    best_f1, best_thr = 0, 0.5
    for thr in np.arange(0.05, 0.95, 0.01):
        preds = (rf_probs[:, i] >= thr).astype(int)
        f1    = f1_score(y_val[:, i], preds, zero_division=0)
        if f1 > best_f1:
            best_f1, best_thr = f1, thr
    best_thr_rf[target] = round(best_thr, 2)
    tuned_f1_rf[target] = best_f1
    print(f'  {target:<25} thr={best_thr:.2f}  F1={best_f1:.4f}')

print(f'\nRF Tuned Macro-F1 : {np.mean(list(tuned_f1_rf.values())):.4f}')
print(f'Baseline was      : 0.4169')
```
